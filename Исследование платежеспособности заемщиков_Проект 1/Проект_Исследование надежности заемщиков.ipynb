Другой столбец с пропущенными значениями — `total_income` — хранит данные о доходах. На сумму дохода сильнее всего влияет тип занятости, поэтому заполнить пропуски в этом столбце нужно медианным значением по каждому типу из столбца `income_type`. ### Категоризация данных

**Задание 16. На основании диапазонов, указанных ниже, создадим в датафрейме `data` столбец `total_income_category` с категориями:**

- 0–30000 — `'E'`;
- 30001–50000 — `'D'`;
- 50001–200000 — `'C'`;
- 200001–1000000 — `'B'`;
- 1000001 и выше — `'A'`.


**Например, кредитополучателю с доходом 25000 нужно назначить категорию `'E'`, а клиенту, получающему 235000, — `'B'`. Используйте собственную функцию с именем `categorize_income()` и метод `apply()`.**

```python
def categorize_income(income):
    try:
        if 0 <= income <= 30000:
            return 'E'
        elif 30001 <= income <= 50000:
            return 'D'
        elif 50001 <= income <= 200000:
            return 'C'
        elif 200001 <= income <= 1000000:
            return 'B'
        elif income >= 1000001:
            return 'A'
    except:
        pass
```

```python
data['total_income_category'] = data['total_income'].apply(categorize_income)
```

#### 3.1 Есть ли зависимость между количеством детей и возвратом кредита в срок?

Для определения *зависимости между количеством детей и возвратом кредита в срок*, нам понадобятся два показателя:

- `children` — количество детей в семье заемщика;
- `debt` — имел ли заемщик задолженность по возврату кредитов (`0` - нет, `1` - да).

Для повышения достоверности выводов, полученных из настоящего анализа, предлагается:
- добавить еще один показатель `overdue_loan_percentage` - *процент просроченной задолженности по кредиту по всем заемщикам в %*, который будет использоваться во всех заданиях, где нужно проследить зависимость между заданной группой заемщиков и уплатой кредитов в срок. При дальнейшей группировке заемщиков на требуемые категории, станет хорошо видно, какой процент просрочки среди всех платежей имеет та или иная анализируемая группа заемщиков, следовательно, данный показатель даст более четкую картину наличия или отсутсвия зависимости между заданной категорией заемщиков и возвратом кредита в срок;
- разбить всех заемщиков на группы, в зависимости от количества детей в семье и результаты добавить в новый столбец `children_group` - *статус семьи заемщика, в зависимости от количества детей*.

```python
#добавляем к исходной таблице data столбец overdue_loan_percentage 
data['overdue_loan_percentage'] = (data['debt'] / data['debt'].count()) * 100
```

Проверяем, добавился ли новый столбец `overdue_loan_percentage`в общий датафрейм:

```python
data.info()
```

Видим, что в строке с индексом 14 появился новый столбец `overdue_loan_percentage`.  Переходим к следующему этапу.               

С помощью функции `children_group()` разобьем всех заемщиков на группы, в зависимости от количества детей  в семье:

```python
def children_group(children):
    """
    Возвращает статус семьи по значению количества у заемщика детей, используя правила:
    - 'бездетная семья' при значении children == 0;
    - 'малодетная семья' при значении children до двух человек;
    - 'многодетная семья' - во всех остальнных случаях (при значении children от 3 и выше)
    """
    if children == 0:
        return 'бездетная семья'
    
    if children <= 2:
        return 'малодетная семья'
    
    return 'многодетная семья'

# проверяем, правильно ли отвечает функция на переданные ей числа
children = 0
print(children_group(children))
children = 1
print(children_group(children))
children = 3
print(children_group(children))

# доавляем новый столбец children_group в data
data['children_group'] = data['children'].apply(children_group)

# выводим для проверки первые 10 строк столбца data['children_group']
print(data['children_group'].head(10))
```

Теперь сформируем по столбцам, необходимым нам для анализа ( `debt`, `overdue_loan_percentage`, `children_group`) сводную таблицу **data_pivot_children_debt** методом *pivot_table()*:

```python
data_pivot_children_debt = data.pivot_table(index=['children_group'],\
                                            values = ['debt', 'overdue_loan_percentage'] , aggfunc='sum')

print() # пробел для лучшей визуалиции данных при выводе на экран
print('Анализ наличия зависимости между количеством детей и возвратом кредита в срок:')
display(data_pivot_children_debt)

#выведем на экран еще общее число заемщиков
print('Итого заемщиков:', data['children'].count())
```

Согласно данным сводной таблицы `data_pivot_children_debt` прослеживается **обратная пропорциональная зависимость между количеством детей и возвратом кредита в срок**. 

Заемщики, которые в анкете указали на отсутвие детей, намного чаще имеют просроченную задолженность по кредиту, чем заемщики с детьми. Причем, процент невозврата кредита в срок напрямую зависит от количества детей должника:

А если более детализировать список заемщиков по количеству детей (от 0 до 5), то можно увидеть, что у заемщиков с **пятью детьми** совсем **нет просрочек** по выплатам кредитов:

```python
#Cоздаем сводную таблицу `data_pivot_famstatus_debt` с данными о семейном положении заемщика и просрочках:
data_pivot_famstatus_debt = data.pivot_table(index=['family_status'], \
                                             values=['debt', 'overdue_loan_percentage'], aggfunc='sum')

#Применим к data_pivot_famstatus_debt (столбцу debt) сортировку по убыванию (ascending = False):
data_pivot_famstatus_debt = data_pivot_famstatus_debt.sort_values(by='debt', ascending=False)
```

Для более полного анализа заведем переменную `debt_sum` с **суммой всех случаев просрочки кредитов** 
по категориям столбца `family_status`: 

```python
debt_sum = data_pivot_famstatus_debt['debt'].sum()
```

Рассчитаем величину просрочек у групп заемщиков в столбце `family_status`, исключая доминирующую категорию (максимальную величину) и долю просрочек в преобладающей категории "женат / замужем". Это необходимо для дальнейшего сравнения и помощи в формировании более развернутых и содержательных выводов.

```python
# рассчитаем сумму всех просрочек по категриям в "family_status", исключая максимальную величину:
debt_difference = debt_sum - data_pivot_famstatus_debt['debt'].max()

# найдем долю просрочек в преобладающей категории "женат / замужем" в %:
share_debt = (data_pivot_famstatus_debt['debt'].max() / debt_sum) * 100

print()
print('Анализ наличия зависимости между семейным положением и возвратом кредита в срок:')

display(data_pivot_famstatus_debt) 
print()
print('Итого просрочек по категориям "family_status":',debt_sum)
print('Сумма просрочек за исключением "женат / замужем":', debt_difference )
print('Доля просрочек категории "женат / замужем":', share_debt )
```

Как видно из таблицы *data_pivot_famstatus_debt* **заемщики в статусе "женат / замужем"** на `3.96%`  **чаще просрочивают платежи по кредитам** по сравнению с заемщиками, находящимися в **разводе** (процент просрочки по кредиту - 0.39%) и на `4.05%` по сравнению с заемщиками в статусе **"вдовец / вдова"**(процент просрочки по кредиту - 0.30%)

То есть, общее количество заемщиков с просроченной кредиторской задолженностью, находящихся в официальном браке, превышает суммарное количество всех остальных категорий заемщиков: 

- женат / замужем - **927 случаев** просрочки;

- остальные категории family_status - **805 случаев** просрочки.

**Удельный вес просрочек категории "женат / замужем"** в отношении всех просрочек по категориям "family_status" -  **53.52%**.


*Даже если сгруппировать всех заемщиков категории "family_status" по группам*: "в браке", "холост", "вдова/вдовец" - ситуация не изменится. **Зависимость просроченных платежей от статуса семейного положения сохранится**:


```python
# c помощью функции family_status_group(family_status) распределим всех заемщиков по группам (семейному положению)
def family_status_group(family_status):
    """
    Возвращает статус семьи по значению family_status заемщика, используя правила:
    - 'в браке' при значении family_status -  'женат / замужем' или 'гражданский брак' ;
    - 'холостые' при значении family_status - 'Не женат / не замужем' или 'в разводе';
    - 'вдовец / вдова' - во всех остальнных случаях.
    """
    if family_status == 'женат / замужем' or family_status == 'гражданский брак':
        return 'в браке'
    
    if family_status == 'Не женат / не замужем' or family_status == 'в разводе':
        return 'холостые'
    
    return 'вдовец / вдова'

# проверяем, правильно ли отвечает функция на переданные ей данные
family_status = 'Не женат / не замужем'
print(family_status_group(family_status))


# доавляем новый столбец family_status_group в data
data['family_status_group'] = data['family_status'].apply(family_status_group)

# Проверяем, добавился ли новый столбец family_status_group общий датафрейм
print(data.info())
```

#### 3.3 Есть ли зависимость между уровнем дохода и возвратом кредита в срок?

Чтобы определить, есть ли зависимость между уровнем дохода и возвратом кредита в срок, сгруппирем всех заемщиков в категории по уровню дохода, используя столбец `total_income` — ежемесячный доход.

При группировке заемщиков, будем использовать критерии по доходам, приведенные в монографии социологического центра РАН в 2023 году ([Статья](https://ngs.ru/text/economics/2023/07/05/72462857/) от 5 июля 2023 г.):

1. **Наименее обеспеченные заемщики** - ежемесячный доход: **менее 1 прожиточного минимума** / 15 453 руб.;
2. **Низкообеспеченные заемщики** - ежемесячный доход: **1-3 прожиточного минимума** / от 15 453 руб. до 46 359 руб.;
3. **Среднеобеспеченные заемщики** - ежемесячный доход: **3-11 прожиточных минимумов** / от 46 359 руб. до 169 983  руб.;
4. **Высокообеспеченные заемщики** - ежемесячный доход: **более 11 прожиточных минимумов**/ выше 169 983  руб.

*Примечание: с 1 января 2024 г. **величина прожиточного минимума** в целом по РФ на душу населения установлена в размере **15 453 руб**.*

C помощью функции `family_status_group(family_status)` распределим всех заемщиков по группам, в зависимости от величины ежемесячного дохода:

```python
def income_group(total_income):
    """
    Возвращает статус заемщика по значению его total_income, используя правила:
    - 'Наименее обеспеченные' при значении total_income < 15 453 руб. ;
    - 'Низкообеспеченные' при значении total_income от 15 453 руб. до 46 359  руб.;
    - 'Среднеобеспеченные' при значении total_income от 46 359 до 169 983 руб.
    - 'Высокообеспеченные' во всех остальнных случаях (выше 169 983).
    """
    if total_income < 15453:
        return 'Наименее обеспеченные'
    if total_income <= 46359:
        return 'Низкообеспеченные'
    if total_income <= 169983:
        return 'Среднеобеспеченные'
    
    return 'Высокообеспеченные'

# проверяем, правильно ли отвечает функция на переданные ей данные
total_income = 35000
print(income_group(total_income))
total_income = 98000
print(income_group(total_income))

#добавляем новый столбец children_group в data
data['income_group'] = data['total_income'].apply(income_group)

# выводим для проверки первые 10 строк столбца data['income_group']
print()
print(data['income_group'].head())
```

Теперь, когда все заемщики распределены на группы в зависимости от величины ежемесячного дохода, создаем сводную таблицу `data_pivot_income_debt`. Для анализа, как и прежде, будем использовать данные из столбцов `'debt'`, `'overdue_loan_percentage'` и созданного `'income_group'`:

```python
data_pivot_income_debt = data.pivot_table(index=['income_group'], \
                                             values=['debt', 'overdue_loan_percentage'], aggfunc='sum')

# отсортируем в data_pivot_income_debt по убыванию столбец 'debt' (ascending = False):
data_pivot_income_debt = data_pivot_income_debt.sort_values(by='debt', ascending=False)

print()
print('Анализ наличия зависимости между уровнем дохода и возвратом кредита в срок:')
display(data_pivot_income_debt)
```

Из таблицы `data_pivot_income_debt` видно, что **самый высокий уровень риска просрочки платежей характерен для среднеобеспеченных клиентов**: `1141` случай просрочки на `574` случая у высокообеспеченных и `17` случаев у низкообечпеченных заемщиков.

*Низкообеспеченные клиенты имеют **наименьший потенциал риска несвоевременной уплаты платежей** по кредиту*.

Как ни странно, но *клиенты, попадающие в категорию высокообеспеченных*, так же **имеют достаточно высокую вероятность просрочки платежей**. Процент несвоевременной выплаты у этой группы составляет -  `2,69%`.

#### 3.6 Объясните, почему заполнить пропуски медианным значением — лучшее решение для количественных переменных.

Перед тем, как объяснить, почему пропуски для количественных переменных лучше заполнять медианным значением, необходимо напомнить, что в целом **пропуски заполняются либо медианой, либо средним арифметическим значением**.

**Отличие медианного значения от среднего арифметического в способе нахождения характерного значения**.

Если **среднее арифметическое** — это сумма всех значений, поделённая на количество значений, то **медиана** — это такое число в выборке, что ровно половина элементов больше него, а другая половина — меньше.

Заполнение пропусков в колличественных значениях медианой наиболее предпочтительно, так как среднее значение некорректно характеризует данные, в особенности, когда некоторые показатели сильно выделяются среди большинства. Происходит завышение/занижение показателей (в зависимости от ситуации), и как следствие, искажение данных.



В задании 7 мы для каждого типа занятости выводили **медианное значение** трудового стажа `days_employed`:

```python
print(data.groupby('income_type')['days_employed'].median())
```

Напомним,что у двух типов (безработные и пенсионеры) получились аномально большие значения. Исправить такие значения сложно, поэтому мы оставили их как есть.

Теперь попробуем для каждого типа занятости вывести **среднее арифметическое значение** трудового стажа `days_employed`:

```python
print(data.groupby('income_type')['days_employed'].mean())
```

У следующих категорий: `госслужащий`, `компаньон`, `сотрудник` среднее арифметического значение трудового стажа больше, чем медианное значение.  Это подтверждает вышесказанное и говорит о том, что среднее арифметическое в количественных значениях существенно искажает распределение.

Теперь попробуем этим же способом для каждого типа занятости вывести и сравнить медиану и среднее значение уровня ежемесячного дохода.

**Медианное значение** дохода `total_income`:

```python
print(data.groupby('income_type')['total_income'].median())
```

**Среднее арифметическое** значение дохода `total_income`:

```python
print(data.groupby('income_type')['total_income'].mean())
```

К категориям, у которых изменились показатели при расчете среднего значения: `госслужащий`, `компаньон`, `сотрудник`, добавилась еще одна - `пенсионер`. 
Как и при расчете показателей стажа, среднее значение уровня дохода выше, чем медиана. 
Это вновь говорит о том, что медиана, в отличие от среднего значения, отдает более точные значения. А это, в свою очередь, дает возможность аналитику работать с наиболее реальными данными, что улучшает качество подготовленного им, по результатам анализа, заключения.

***Таким образом, можно сделать вывод, что заполнение пропусков медианным значением — лучшее решение для количественных переменных. Изменения в данных имеют меньший эффект на медиану, что придает ей стабильность и защищает от воздействия случайных факторов или изменений.***
