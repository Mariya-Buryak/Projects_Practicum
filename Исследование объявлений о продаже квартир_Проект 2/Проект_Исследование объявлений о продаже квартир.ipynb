Так как мы проводим анализ именного жилых помещений **из данных следует исключить показатели, ниже 2.1 метра и выше 4 метров**:

````python
# Подсчитаем количество квартир с каждой высотой потолков и отсортируем по убыванию
ceiling_height_counts = apartment_data['ceiling_height'].value_counts().sort_index(ascending=False).head(10)

# Выведем результаты
print(ceiling_height_counts)
````

Скорее всего, это вещественные значения: 2.0 м, 2.5 м, 3.2 м и т.д. Наверное, продавцы не поставили символ разделитель. Исправим эту ситацию.  

````python
# Создадим словарь для замены значений
replace_values = {32.0: 3.2, 27.5: 2.75, 27.0: 2.7, 26.0: 2.6, 25.0: 2.5, 24.0: 2.4, 22.6: 2.26, 20.0:2.0}

# Заменим значения в столбце 'ceiling_height' с использованием словаря
apartment_data['ceiling_height'] = apartment_data['ceiling_height'].replace(replace_values)

# Проверим результаты
print(apartment_data['ceiling_height'].value_counts().sort_index(ascending=False).head())
````

````python
# С помощью функции query() созданим пременную filtered_data, выбрав строки,
#где ceiling_height в диапазоне между 2.1 и 4 метра включительно
filtered_data = apartment_data.query('2.1 <= ceiling_height <= 4')

# Проверяем распределение данных в столбце 'ceiling_height'
filtered_data['ceiling_height'].describe()
````

Отклонение минимально (0.26), медианное значение в заданной выборке, как и в столбце со всеми аномальными значениями `filtered_data['ceiling_height']`, не изменилось и равно **2.65 метров**, им и заполним пропуски в столбце `'ceiling_height'` тaблицы `apartment_data`.

````python
# Заполним пропуски в столбце 'ceiling_height' медианным значением 
apartment_data['ceiling_height'] = apartment_data['ceiling_height'].fillna(apartment_data['ceiling_height'].median())

# Проверяем замену пропусков на медиану 2.65
print(apartment_data['ceiling_height'])

# Проверяем остались ли пропуски в столбце 'ceiling_height'
print('-'*50)
print('Количество пропусков в "ceiling_height" - ', apartment_data['ceiling_height'].isna().sum())
````

*Причиной возникновения дубликатов в наименовании населенных пунктов можно назвать опечатки и различные варианты их написания*.  

#### Во `floors_total` всего 86 пропусков, что составляет лишь 0.36% анализируемой информации.

У нас нет возможности получить наиболее точные данные, поэтому, можно либо оставить как есть, либо заполнить их медианным значением этажности дома по каждому населенному пункту. Чем мы и займемся. 

````python
# Заполняем пропуски медианным значением этажности дома по каждому населенному пункту:
for t in apartment_data['locality_name'].unique():
    apartment_data.loc[(apartment_data['locality_name'] == t) & (apartment_data['floors_total'].isna()), 'floors_total'] = \
    apartment_data.loc[(apartment_data['locality_name'] == t), 'floors_total'].median()
    
print('Количество пропусков в "floors_total" - ', apartment_data['floors_total'].isna().sum())
````

Перевести такие данные в целочисленный тип (int) невозможно, так как int не поддерживает пропущенные значения. Поэтому, сначала переведем в целочисленный тип (int) `last_price`, `floors_total` и`balcony`, не имеющие пропусков. 

````python
# Создаем список conversion_to_int, в который вносим все столбцы, которые нужно преобразовать в int
conversion_to_int = ['last_price', 'floors_total', 'balcony'] 

# Преобразование столбцов в числовой формат
for column in conversion_to_int:
    apartment_data[column] = apartment_data[column].astype('int')
    
# Проверка:
#apartment_data.info()
#apartment_data.head()
````

*К остальным столбцам с пропусками попробовали сначала применить функцию*:

``` python

# Определяем функцию для преобразование столбцов в числовой формат, которая проверяет, является ли значение не NaN, и только тогда преобразует его
def int_value(x):
    if pd.notnull(x):
        return int(x)
    else:
        return x

# Применяем функцию к каждой ячейке выбранных столбцов DataFrame
for column in conversion_to_int:
    apartment_data[column] = apartment_data[column].apply(int_value)
```
НО результат вновь остался тем же: float.

Затем попробовали применить метод `pd.to_numeric()` с параметром `errors="coerce"`. Но результата не было. Потом попробовали заменить NaN на целочисленное значение -1 (можно и на любое другое подходящее целочисленное значение), а затем преобразовать столбцы в целочисленный тип данных, в конце заменив -1 обратно на NaN (это необходимо, чтобы не исказить данные). 

``` python
# Преобразование столбцов в числовой формат с обработкой ошибок и преобразованием непреобразуемых значений в NaN
apartment_data[conversion_to_int] = apartment_data[conversion_to_int].apply(pd.to_numeric, errors='coerce')

# Замена NaN на целочисленное значение
apartment_data[conversion_to_int] = apartment_data[conversion_to_int].fillna(-1).astype('int')

# Замена -1 обратно на NaN
apartment_data = apartment_data.replace(-1, 'NaN')
```

#### Посчитаем среднюю цену одного квадратного метра в 10 населённых пунктах с наибольшим числом объявлений.

````python
# Создаем сводную таблицу
pivot_table_price_for_m2 = filtered_data.pivot_table(index='locality_name', 
                             values=['price_for_m2', 'last_price'], 
                             aggfunc={'price_for_m2': 'mean', 'last_price': 'count'})

# Сортируем по количеству объявлений и выбираем 10 населенных пунктов с наибольшим числом объявлений
top_localities = pivot_table_price_for_m2.nlargest(10, 'last_price')

# Определяем населенные пункты с самой высокой и низкой стоимостью квадратного метра
highest_price_per_m2 = top_localities['price_for_m2'].idxmax()
lowest_price_per_m2 = top_localities['price_for_m2'].idxmin()

# Выводим сводную таблицу и населенные пункты с самой высокой и низкой стоимостью квадратного метра
display(top_localities)
print("\nНаселенный пункт с самой высокой стоимостью квадратного метра:", highest_price_per_m2)
print("Населенный пункт с самой низкой стоимостью квадратного метра:", lowest_price_per_m2)
````

#### Выделим квартиры в Санкт-Петербурге с помощью столбца `locality_name` и вычислим их среднюю стоимость на разном удалении от центра.

Учитывать будем каждый километр расстояния: узнаем среднюю цену квартир в одном километре от центра, в двух и так далее. Построим график изменения средней цены для каждого километра от центра Петербурга.

````python
# Выбираем квартиры в Санкт-Петербурге
spb_apartments = filtered_data[filtered_data['locality_name'] == 'Санкт-Петербург'].copy()

# Группируем данные по округленному расстоянию до центра и вычисляем среднюю цену квартир для каждой группы
mean_price_by_distance = spb_apartments.groupby('distance_to_center')['last_price'].mean()

#print(mean_price_by_distance)

# Рассчитываем коэффициент корреляции Пирсона м/у ценой и удаленностью от центра
correlation = spb_apartments['last_price'].corr(spb_apartments['distance_to_center'])

print("Коэффициент корреляции между last_price и distance_to_center:", correlation.round(2))
````

Отрицательная корреляция (-0.22) говорит о том, что при увеличении одной величины, уменьшается вторая. Т.е. при увеличении расстояния от центра города, стоимость квартир должна снижаться. Предлагается построить графики и наглядно в этом убедиться.

````python
# Строим график зависимости средней цены квартир от расстояния до центра
plt.figure(figsize=(7, 4))
plt.plot(mean_price_by_distance.index, mean_price_by_distance.values, marker='o', linestyle='-')
plt.title('Зависимость средней цены квартир от расстояния до центра Санкт-Петербурга')
plt.xlabel('Расстояние до центра, км')
plt.ylabel('Средняя цена квартиры')
plt.grid(True)
plt.show()

mean_price_by_distance.plot(kind='bar', figsize=(7,4))
plt.xlabel('Километры от центра города')
plt.ylabel('Средняя стоимость предложения')
plt.show()
````

Действительно, квартиры в центре Санкт-Петербурга оцениваются выше по сравнению с более удаленными районами.

Также можно заметить пик между 25 и 29 км. Есть предположение, что в пределах 28 км от центра Санкт-Петербурга находится населенный пункт, в котором цены на жилье выше, чем в других пригородах. Воспользовавшись дополнительными источниками информации, можно предполодить, что это Всеволожск (Всеволожский рн.). Находится на расстоянии примерно 27-28 км от центра Питера, лидер по темпам ввода жилья в Ленинградской области в 2010 году. Во Всеволожске реализуется множество проектов по возведению многоквартирных домов, строятся малоэтажные комплексы. Так же зафиксирован самый низкий уровень безработицы по регионам, много объектов промышленности. Много парков, парковых зон, озер и водоемов. Такие сделки были отнесены нами к *категории быстрых*. 
В категорию *"обычная продажа"* попали сделки сроком до года. После года наблюдается падение объемов продаж недвижимости. К категории *"необычно долгих продаж"* отнесли аномально долгие продажи - более 1.5 лет. В данных есть объявления, "провисевшие" на сайте почти 4.5 года. Вероятной причиной аномально долгих продаж могут быть  какие-нибудь проблемы с квартирой: завышена цена и площадь, нужен ремонт, неудобное расположение, район, есть обременения (ипотека, мат.капитал), несколько собственников, есть невыписанные жильцы - все это требует более бдительного подхода к сделке и влияет на срок оформления сделки, а, возможно, продавец провел оффлайн сделку и забыл снять объявление с публикации на сайте.

Затем **определили факторы, которые больше всего влияют на общую (полную) стоимость объекта**. Рассчитали коэффициенты корреляции для количественных показателей, получили следующие результаты: корреляция в 0.66 говорит о тесной взаимосвязи цены и общей площади, корреляция в 0.53, 0.54 о наличии связи между ценой и жилой площадью и площадью кухни соответственно, однако не слишком сильной. Для категоральных же параметров (день недели, месяц, год публикации объявления, тип этажа и количество комнат)  были построены по средним значениям цены графики типа "барплот", сформулированы следующие выводы:

- связь между ценой и днем недели размещения объявления минимальна. Цены не сильно отличаются друг от друга, с понедельника по четверг наблюдается небольшой рост стоимости, четверг - пиковый. Затем идет спад. В воскресенье наблюдается самая низкая цена на объекты недвижимости;
- в 2014-2015 годах наблюдаются самые высоки цены на жилье. Пиковый - 2014 год. Идет плавное снижение стоимости вплоть до 2018 года. В 2019 году вновь наблюдается рост цен;
- Между ценой на квартиры и количеством комнат есть прямая зависимость. С ростом числа комнат, растет и цена объектов недвижимости. Пиковым являются цены на семикомнатные квартиры. 8-комнатные по ценнику ниже;
- дешевле всего квартиры на первом этаже. Последний этаж, как ни странно, пользуется спросом. Цена на квартиры на последнем этаже немного ниже стоимости промежуточных квартир.

Далее посчитали среднюю цену одного квадратного метра в 10 населённых пунктах с наибольшим числом объявлений и выделили населённые пункты с самой высокой и низкой стоимостью квадратного метра. **Самая высокая стоимостью квадратного метра в Санкт-Петербург (111 022 руб/м2), самая низкая - в Выборге (57272 руб/м2.)**. Круговой диаграммой наглядно показали, что наиболее часто продаются квартиры в Санкт-Петербурге и его пригороде - г. Пушкине, деревне Кудрово, пос. Парголово, пос. Мурино (которые граничат вплотную с Санкт-Петербургом, можно добраться на метро). В Выборге меньше всего объявлений о продаже недвижимости. Возможно, здесь играет роль еще местоположение - до Выборга ехать целых 138 км. 

Затем выделили квартиры в Санкт-Петербурге с помощью столбца `locality_name` и вычислили их среднюю стоимость на разном удалении от центра. Было выявлено, что **с увеличением расстояния от центра города стоимость квартир снижается (отрицательная корреляция), т.е. квартиры в центре Санкт-Петербурга оцениваются выше по сравнению с более удаленными районами**.  Пик между 25 и 29 км объяснили тем, что в 28 км от центра Санкт-Петербурга находится населенный пункт, в котором цены на жилье выше, чем в других пригородах. Воспользовавшись дополнительными источниками информации, предположили, что это Всеволожск (Всеволожский рн.). Находится на расстоянии примерно 27-28 км от центра Питера, лидер по темпам ввода жилья в Ленинградской области в 2010 году. Во Всеволожске реализуется множество проектов по возведению многоквартирных домов, строятся малоэтажные комплексы. Так же зафиксирован самый низкий уровень безработицы по регионам, много объектов промышленности. Много парков, парковых зон, озер и водоемов. Может все эти факторы и сыграли на ценообразовательной политике, выделив пик на расстоянии от центра 27 км.
