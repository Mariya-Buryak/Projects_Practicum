Создадим список `categories` и поместим в него эти столбцы, а затем выведем все уникальные значения, чтобы понять, какая общая проблема встречается почти во всех категориальных столбцах.

````python
# Создаем список, помещаем туда все категоральные значения
categories = ['title', 'type', 'film_studio', 'production_country', 'director', 'producer', 'age_restriction', 'genres']

# С помощью цикла выводим на экран все уникальные значения, разделяем "-----"
for category in categories:
    unique_values = rus_film_distribution[category].unique()
    #print('-'*35)
    # print(f"Уникальные значения в столбце '{category}': {unique_values}")
````

**Общая проблема, встречающаяся почти во всех категориальных столбцах, заключается в том, что уникальные значения содержат пробелы или другие символы, а также содержат избыточное количество уникальных значений из-за перечисления в них отдельных признаков**. 

Например, в столбце `type` есть значения с пробелами в начале строки, а в столбце `production_country` значения содержат разделительные символы, такие как запятые или тире, в столбце genres в одной строке по 2-3 жанра.

Эти проблемы могут повлиять на корректную обработку данных и требуют нормализации данных.

Удалим лишние символы, пробелы и разделители.

````python
# Создаем функцию для очистки значений в категориальных столбцах
def clean_categorical_columns(df, columns):
    for col in columns:
        # Удаляем лишние пробелы
        df[col] = df[col].str.strip()
        # Удаляем двойные запятые
        df[col] = df[col].str.replace(',,', ',')
        # Удаляем пробелы в начале и конце строки
        df[col] = df[col].str.strip()
    return df

# Применим функцию к DataFrame
rus_film_distribution = clean_categorical_columns(rus_film_distribution, categories)

# Отдельно заменим разделители символов в production_country на запятые
rus_film_distribution['production_country'] = \
rus_film_distribution['production_country'].str.replace('-', ',').str.replace(';', ',')

# И удалим пробелы внутри строк в столбце 'production_country'
rus_film_distribution['production_country'] = \
rus_film_distribution['production_country'].str.replace(' ', '')

# Выведем для проверки уникальные значения столбца type
set(rus_film_distribution['type'].unique())
````

Следовательно, должны быть отрицательные величины в разнице между суммой господдержки и величиной бюджета. Давайте проверим эту теорию.

Для удобства и сохранности данных, создадим срез данных `budget_check` по столбцам `['refundable_support', 'nonrefundable_support', 'budget']`. Поскольку Pandas обычно выдает предупреждение при добавлении новых столбцов в срез данных, чтобы избежать этого предупреждения, включим в код следующую строку.:

```python 
pd.options.mode.chained_assignment = None
```

Затем создадим новый столбец `total_support` с суммой значений `refundable_support` и `nonrefundable_support`, удалим для удобства строки с пропущенными значениями в столбцах `refundable_support`, `nonrefundable_support` и `budget` - чтобы визуально посмотреть на строки с имеющимися данными и понять, есть ли с ними какие-нибудь проблемы.

````python
pd.options.mode.chained_assignment = None

budget_check = rus_film_distribution[['refundable_support', 'nonrefundable_support', 'budget']]

# Создаем новый столбец с суммой значений refundable_support и nonrefundable_support
budget_check['total_support'] = budget_check['refundable_support'] + budget_check['nonrefundable_support']

# Удалим строки с пропущенными значениями 
budget_check = budget_check.dropna(subset=['refundable_support', 'nonrefundable_support', 'budget'])

# Создадим столбцы для разницы между суммой поддержки и бюджетом
budget_check['difference'] = budget_check['total_support'] - budget_check['budget']

# Создадим так же столбец с долей "недостачи" средств
budget_check['lack_funds'] = (100 - (budget_check['total_support'] / budget_check['budget'] *100)).round(2)

# Выведем первые 30 строк для анализа
display(budget_check[['refundable_support', 'nonrefundable_support', 'total_support', 'budget', 'difference', 'lack_funds']].head(40))
````

#### Создадим два столбца: с именем и фамилией главного режиссёра `director_fullname` и основным жанром фильма `main_genre`. В столбцы войдут первые значения из списка режиссёров и жанров соответственно.

Чтобы создать такие столбцы, лучше всего использовать собственную функцию. Эту функцию можно применить к двум столбцам сразу. 

````python
# Создаем функцию extract_first_value
def extract_first_value(row, column_name):
    value = row[column_name] #ф-ия извлекает значение из указанного столбца column_name в текущей строке row
    if isinstance(value, str): #проверяет, является ли значение строкой
        return value.split(',')[0]  #разделяет строку по запятой и возвращает первый элемент списка
    else:
        return None

# Создаем столбец с именем и фамилией режиссера director_fullname
rus_film_distribution['director_fullname'] = rus_film_distribution.apply(extract_first_value, 
                                                                         args=('director',), axis=1)

# Создаем столбец с основным жанром фильма main_genre
rus_film_distribution['main_genre'] = rus_film_distribution.apply(extract_first_value, 
                                                                  args=('genres',), axis=1)

# Проверка
print(rus_film_distribution['director_fullname'].head())
print(rus_film_distribution['main_genre'].head())
````

````python
# Создаем функцию extract_age_group
def extract_age_group(row, column_name):
    value = row[column_name] #ф-ия извлекает значение из указанного столбца column_name в текущей строке row
    if isinstance(value, str): #проверяет, является ли значение строкой
        return value.split(' - ')[0]  #разделяет строку по дефису и возвращает первый элемент списка
    else:
        return None

# Создаем столбец с основным жанром фильма main_genre
rus_film_distribution['age_group'] = rus_film_distribution.apply(extract_age_group, 
                                                                  args=('age_restriction',), axis=1)

# Проверка
print(rus_film_distribution['age_group'].head())
````

````python
# В positive_difference запишем все окупившиеся кинокартины
positive_difference = return_support_pivot[return_support_pivot['return_support'] > 0]

# В negative_difference запишем все не окупившиеся кинокартины
negative_difference = return_support_pivot[return_support_pivot['return_support'] <= 0]

print('Фильмы, окупившие гос.поддержку :', positive_difference.shape[0])
print('Фильмы, не окупившие гос.поддержку:', negative_difference.shape[0])
print('% не окупившихся фильмов с гос.поддержкой:', (positive_difference.shape[0] / negative_difference.shape[0]) * 100)
````

Видим, что результат совсем не утешительный - **59% фильмов с государственной поддержкой не окупились**. Вычислим общую сумму государственной поддержки для не окупившихся и окупившихся фильмов:

````python
# Общая сумма гос.поддержки для не окупившихся фильмов
total_support_non_profitable = negative_difference['total_support'].sum()
print("Сумма государственной поддержки для не окупившихся фильмов:", total_support_non_profitable)

# Общая сумма гос.поддержки для окупившихся фильмов
total_support_profitable = positive_difference['total_support'].sum()
print("Сумма государственной поддержки для окупившихся фильмов:", total_support_profitable)
````

**Сумма государственной поддержки для не окупившихся фильмов** за весь анализируемый период (с 2010 по 2019 гг.) **составила: 9 283 505 860 руб.**, для окупившихся - 10 027 662 452 руб. Радует, что у окупившихся фильмов сумма гос.поддержки, хоть и ненамного, но больше.

**Провели исследовательский анализ данных**.

Определили, сколько данных о фильмах, выходивших в прокат каждый год, содержат сведения о денежных средствах, полученных от кинопроката. Согласно графикам, **в 2010 году меньше всего информации о сборах от проката кинокартин**. Всего в этот год в прокате было 985 фильмов, но суммы от кассовых сборов отражены только для 101 фильма. До 2014 года наблюдается плавный рост сумм сборов. С 2015 доля фильмов со сборами уже не ниже 50%. **Пик зарегистрирован в 2017 году**. Именно в этот период было меньше всего премьер за весь анализируемый период, но данные о суммах сборов от проката представлены полнее всего: 357 фильмов из 505 указаны с суммой кассовых сборов.

**Больше всего в прокате было фильмов в 2010 году - 985 премьер**. Потом были замечены скачки снижения/роста числа премьер. В 2019 году количество вновь выросло и составило 930 фильмов за год. В 2010 году сумма сборов была минимальной, в 2018 году - максимальной, 2 428 654 руб. и 49 668 403 134 руб., соответсвенно.

Также посчитали среднюю и медианную сумму сборов для каждого года. **За весь период выборки (в особенности, с 2013 года) наблюдается стабильный рост суммы сборов за прокат фильмов**. В течение этого периода общая сумма сборов продолжала увеличиваться. Несмотря на значительные изменения в средних значениях за эти годы, медианное значение суммы сборов оставалось относительно постоянным. Разница между медианным и средним значениями суммы сборов значительна и координально начала "показывать себя" с 2014 года.

Далее определили, что **меньше всего денег собрали фильмы/мультики без возрастного ограничения**, т.е. «0+» (для любой зрительской аудитории). Сумма сборов фильмов 0+ с 2015 по 2019 год имела тенденцию постепенного снижения. Видимо, родители с маленькими детьми редко ходят в кино, но это абсолютно нормально. Ребенок в возрасте до 4 лет вообще не усидчив и редко может высидеть весь сеанс.

Следующими по сумме сборов выступают фильмы с возрастным ограничением 18+. Здесь тоже все логично. **Возрастное органичение значительно уменьшает количество целевой аудитории**. У тому же, можно предположить, что фильмов 18+ значительно меньше выходит на экран из-за того, что такие экранизации могут получать меньшее количество прокатных копий по причине ограничений на показ в кинотеатрах и цензуры. Однако, посмотрев на количественное распределение премьер, увидели, что фильмы 18+ и 16+ делят с собой первенство по количеству премьер в разрезе каждого года. Значит, дело именно в специфичности жанра, вкусовых предпочтениях и ограничения потенциальной аудитории. Но важно отметить тенденцию роста собранных с просмотров сумм. В 2019 году наблюдается самый высокий показатель.

**Фильмы с ограничениями 6+ с 2015 по 2019 гг. собирали больше денег, чем фильмы 18+**. Видимо, очень много родителей с детьми и молодежи ходили на премьеры мультиков и сказок. В 2015-2016 гг. наблюдается постепенный рост притока средств, с 2016 сборы плавно снижались вплоть до 2018 г., в 2019 году заметно повышение сумм кассовых сборов - он же явлеяется и лидером по полученным суммам средств.

**Фильмы с возрастным ограничением 12+ в 2015 году занимали второе место по кассовым сборам**. С 2015 по 2017 год наблюдается  резкий спад притока средств; а 2017 год - пиковый по самым маленьким сборам за весь анализируемый период. Затем в 2018 г. вновь резкий рост, однако, к 2019 году вновь объем полученных денежных средств от проката фильмов 12+ снизился. В 2018 году собрано больше всего денег от проката.

**Лидерами по собранным денежным средствам от проката фильмов выступают премьеры с возрастной категорией 16+**. На протяжении всего анализируемого периода суммы от проката занимали либо лидирующее положение, либо делили первенство с другой категорией: в 2015 больше собрали только от проката фильмов 12+, в 2019 - от фильмов 6+. С 2015 по 2017 год набллюдался стабильный рост поступлений средств от проката, но с 2017 такое же стабильное снижение сборов.

**В 2019 году сумма полученных денег от проката кинокартин практически сравнялись почти для всех возрастных категорий - в одной точке встретились суммы поступлений от проката фильмов 6+, 12+ и 16+**.

**Причина резкого снижения в 2017 году сумм сборов от проката фильмов категории 12+ и пик платежей у фильмов 16+ может быть свзан со следующей причиной**: как уже говорилось ранее, в 2017 году, по сравнению с двумя предыдущими годами, был резкий спад количества прокатных фильмов. в 2017 году фильмов с возрастным ограничением 12+ было всего 41, а с возрастным ограничением 16+ - 121, в три раза больше. Соответвенно, сборы от одной анализируемой категории резко снизились, а от второй,наоборот, выросли.

С помощью диаграммы размаха **составлен рейтинг фильмов по группам возрастных ограничений максимального размера кассовых сборов от кинопроката**: Фильмы с ограничением 12+, Фильмы с ограничением 6+, Фильмы с ограничением 16+, Фильмы с ограничением 18+, Фильмы с ограничением 0+.

Таким образом, величина кассовых сборов прямым образом зависит от возрастных ограничений, а так же количество кинокартин, выпускаемых в прокат. Ситуация меняется в зависимости от года. На это влияет множество факторов (какие картины в прокате, финансовое положение зрителей, возможность посмотреть фильмы у себя дома и т.д). Но, в целом, можно сдлать вывод, что к концу анализируемого периода наблюдается прирост средств по сравнению с предыдущими периодами.
